```python
# map을 사용하여 정수로 변환하기
a, b = map(int, ["10", "20"])
print(a)
print(b)
print("=========")
print(type(a))
print(type(b))
```

```python
a, b = map(int, input("숫자 두 개를  입력하세요:").split())
print(a + b)
```

```python
# 입력받은 값을 콤마를 기준으로 분리하기
a, b = map(int, input("숫자 두 개를  입력하세요:").split(","))
print(a + b)
```

```python
# 연습문제
a, b, c = map(int, input().split())
print(a + b + c)
```
