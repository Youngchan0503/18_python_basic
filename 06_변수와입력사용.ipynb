```python
x = y = z = 10
print(x)
print(y)
print(z)
# 위에서는 x, y, z가 같은 10의 주소를 가르키지만
# 여기서는 z가 10의 주소를 y가 z의 주소를 x가 y의 주소를 가르킨다
```
