```python
a_class = {'김철수', '이영희', '박민수', '최지영', '홍길동'}
b_class = {'박민수', '최지영', '정수민', '강호동', '이영희'}


# 1. 두 반 모두에 있는 학생 (교집합)
print(a_class & b_class)

# 2. 두 반을 합친 전체 학생 (합집합)
print(a_class | b_class)

# 3. A반에만 있는 학생 (차집합)
print(a_class - b_class)

# 4. B반에만 있는 학생 (차집합)
print(b_class - a_class)
```

```python
# ### 문제 7-1: 택시 요금 계산기

# 거리(km)와 시간대를 입력받아 요금을 계산하세요.

# - 기본요금: 3000원 (2km까지)
# - 추가요금: 2km 초과시 1km당 1000원
# - 심야할증(22시~04시): 총 요금의 20% 추가
#   distance = float(input("거리를 입력하세요 (km): "))
#   hour = int(input("현재 시간을 입력하세요 (0-23): "))
#   **예상 출력:**
#   거리를 입력하세요 (km): 5
#   현재 시간을 입력하세요 (0-23): 23
#   요금: 7200원

distance = float(input("거리를 입력하세요 (km): "))
hour = int(input("현재 시간을 입력하세요 (0-23): "))

charge = 3000

if distance > 2:
    charge = charge + (distance - 2) * 1000


if 22 <= hour or hour <= 4:
    charge = charge * 1.2

print(f"요금: {charge}원")

```
