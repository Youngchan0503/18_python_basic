```python
# 사용자가 입력한 값에 if 조건문 사용하기
x = int(input())

if x == 10:
    print("10입니다.")
if x == 20:
    print("20입니다.")
```

```python
# 심사문제
price = int(input())
coupon = input()

if coupon == "Cash3000":
    print(price - 3000)
if coupon == "Cash5000":
    print(price - 5000)

```

```python

# 만약 쿠폰의 이름에 할인 가격 형태가 적용되는 형태라면 if 없이 가능
price = int(input())
coupon = input()

discount = coupon[4:]
print(type(discount))
print(price - int(discount))
```
