```python
# else 사용하기

# if 조건식:
#     수행할코드
# else:
#     수행할코드

x = 5
if x == 10:
    print("10입니다")
else:
    print("10이 아닙니다")
```

```python
# if 조건문의 동작 방식 알아보기
if True:
    print("참")
else:
    print("거짓")
    
if False:
    print("참")
else:
    print("거짓")
    
if None:
    print("참")
else:
    print("거짓")
    
    
```

```python
# 조건문에 숫자 지정하기

if 0:
    print("참")
else:
    print("거짓")

if 1:
    print("참")
else:
    print("거짓")

if 0x1F:
    print("참")
else:
    print("거짓")

if 0b1000:
    print("참")
else:
    print("거짓")

if 13.5:
    print("참")
else:
    print("거짓")
```

```python
# if 조건문에 문자열 지정하기
if "hello":
    print("참")
else:
    print("거짓")

if "":
    print("참")
else:
    print("거짓")

if " ":
    print("참")
else:
    print("거짓")
```

```python
# 조건식을 여러개 지정하기
x = 10
y = 20

if x == 10 and y == 20:
    print("참")
else:
    print("거짓")
```

```python
# 심사문제
korean, english, math, sience = map(int,input().split())

if 0 <= korean <= 100:
    if 0 <= english <= 100:
        if 0 <= math <= 100:
            if 0 <= sience <= 100:
                avg = (korean + english + math + sience) / 4
                if avg >= 80:
                    print("합격")
                else:
                    print("불합격")
            else:
                print("잘못된 점수")
        else:
            print("잘못된 점수")
    else:
        print("잘못된 점수")
else:
    print("잘못된 점수")

```

```python

korean, english, math, sience = map(int,input().split())

if (0 <= korean <= 100) and (0 <= english <= 100) and (0 <= math <= 100) and (0 <= sience <= 100):
    avg = (korean + english + math + sience) / 4
    if avg >= 80:
        print("합격")
    else:
        print("불합격")
else:
    print("잘못된 점수")
```
