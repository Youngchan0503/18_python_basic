```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for x, y in a:
    print(x, y)
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in a:
    for j in i:
        print(j, end=" ")
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
# while 반복문을 두 번 사용하기
a = [
    [10, 20],
    [30, 40],
    [50, 60]
]

i = 0
while i < len(a):
    j = 0
    while j < len(a[i]):
        print(a[i][j], end=" ")
        j += 1
    print()
    i += 1
```

```python
# 심사문제
col, row = map(int,input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        # 지뢰확인, 지뢰가 아닌 경우 continue
        if matrix[i][j] == ".":
            matrix[i][j] = 0

        if matrix[i][j] != "*":
            continue

        for r in range(i-1, i+2):
            for c in range(j-1, j+2):
                if c < 0 or c >= len(matrix[i]) or r < 0 or r >= len(matrix):
                    continue
                if matrix[r][c] == ".":
                    matrix[r][c] = 0
                if matrix[r][c] != "*":
                    matrix[r][c] += 1

for x in matrix:
    for y in x:
        print(y, end="")
    print()
print()
```

```python
## 심사문제
col, row = map(int,input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        # 지뢰확인, 지뢰인 경우 continue
        if matrix[i][j] == "*":
            continue
        else:
            matrix[i][j] = 0

        for r in range(i-1, i+2):
            for c in range(j-1, j+2):
                if c < 0 or c >= len(matrix[i]) or r < 0 or r >= len(matrix):
                    continue
                if matrix[r][c] == "*":
                    matrix[i][j] += 1

for x in matrix:
    for y in x:
        print(y, end="")
    print()
print()
```
