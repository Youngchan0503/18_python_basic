```python
# 람다표현식으로 함수 만들기
def plus_ten(x):
    return x + 10

print(plus_ten(1))
```

```python
# lambda 매개변수들: 식
lambda x: x + 10
```

```python
plus_ten2 = lambda x: x + 10
plus_ten2(2)
```

```python
# 람다 표현식 자체를 호출하기
# (lambda 매개변수들: 식)(인수들)

(lambda x: x + 10)(2)
```

```python
# 람다 표현식 안에서는 변수를 만들 수 없다.
(lambda x: y = 10; x + y)(1)
```

```python
y = 10
(lambda x: x + y)(1)
```

```python
def plus_ten(x):
    return x + 10

list(map(plus_ten, [1, 2, 3]))
```

```python
list(map(lambda x: x + 10, [1, 2 ,3]))
```

```python
# 람다표현식과 map, filter, reduce 함수 활용하기
a = list(range(1, 11))

print(list(map(lambda x: str(x) if x % 3 == 0 else x, a)))

b = lambda x: str(x) if x % 3 == 0 else x
print(list(map(b, a)))
```

```python
# 람다표현식에 if를 적으면 else를 반드시 적어야 한다
lambda x: str(x) if x % 3 == 0 else 1
```

```python
a = list(range(1, 11))

la = lambda x: str(x) if x == 1 else float(x) if x == 2 else x + 10

list((map(la, a)))
```
