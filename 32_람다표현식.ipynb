```python
# 람다표현식과 map, filter, reduce 함수 활용하기
a = list(range(1, 11))

print(list(map(lambda x: str(x) if x % 3 == 0 else x, a)))

b = lambda x: str(x) if x % 3 == 0 else x
print(list(map(b, a)))
```

```python
# 람다표현식에 if를 적으면 else를 반드시 적어야 한다
lambda x: str(x) if x % 3 == 0 else 1
```

```python
# map에 객체를 여러개 넣기
a = [1, 2, 3, 4, 5]
b = [2, 4, 6, 8, 10]
list(map(lambda x, y: x * y, a, b))
```

```python
# filter 사용하기
# filter(함수, 반복가능한 개체)
def f(x):
    return x > 5 and x < 10

a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

# f(8)

list(filter(f,a))
```

```python
a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]
list(filter(lambda x: x > 5 and x < 10, a))
```

```python
list(filter(lambda x: 5 < x < 10, a))
```

```python
# reduce 사용하기
from functools import reduce
# reduce(함수, 반복가능한 객체)

# import random
# random.randint()

# from random import randint
# randint(1, 10)

def f(x, y):
    return x + y

a = [1, 2, 3, 4, 5]
reduce(f, a)
```

```python
a = [1, 2, 3, 4, 5]
from functools import reduce

reduce(lambda x, y: x + y, a)
```
