```python
## 8진수와 16진수
a = 0o177
print(a)

print(1 * (8 ** 2) + 7 *(8 ** 1) + 7 * (8 ** 0))

b = 0x8ff
print(b)

print(8 * (16 ** 2) + 15 * (16 ** 1) + 15 * (16 ** 0))
```

```python
b = 0xabc
print(b)

print(10 * (16**2) + 11 * (16 ** 1) + 12)
print(0xa)
```

```python
## 사칙연산
a = 3
b = 4


print(a + b)
print(a - b)
print(a * b)
print(a / b)
```
