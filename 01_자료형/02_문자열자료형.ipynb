```python
## 문자열 연산하기
a = "python"
b = " is fun"

print(a)
print(a * 3)

print(a + b)
# print(a - b) 
```

```python
## 문자열 길이 구하기

a = "Life is too short"
len(a)
```

```python
## 문자열 인덱싱과 슬라이싱
a = "Life is too short"
a[3]
a[16]
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a = "Life is too short"
a[0:3+1]
a[5:7]
a[-5:-1]
a[-5:]
a[:4]
```

```python
## 문자열 포매팅
print("I eat %d apples" % 3)
```

```python
print("I eat %s apples" % "five")
```

```python
a = 3
b = "five"
print("I eat %d apples" % a)
print("I eat %s apples" % b)
```

```python
## format() 함수를 사용한 포매팅
"I eat {0} apples".format(3)
"I eat {0} apples".format("five")
```

```python
## 2개 이상의 값을 넣을 때
number = 10
day = "three"

"I ate {0} apples, so i was sick for {1} days".format(number, day)
```

```python
## f문자열 포매팅
name = "홍길동"
age = 30

print(type(name))
print(type(age))

# 앞으로는 이거 주로 쓸 예정
print(f"나의 이름은 {name}이고, 나이는 {age}입니다")
```

```python
## 문자열 관련 함수들
a = "hobby"
type(a)

## 문자 개수 세기 - count()
## . 뒤에 나오는 함수를 메서드라고 함
## 바로 쓸수 있는거는 내장함수
a.count("b")
"hobby".count("b")
```

```python
## 위치 알려주기 - find()
a = "Python is the best choice"

## 인덱스 알려주는 메서드(제일 처음 나오는 인덱스만 반환하는거 같음, 대소문자 구분함, 없으면 -1 반환)
print(a.find("b"))
print(a.find("k"))
```

```python
## 위치 알려주기 - index()
a = "Life is too short"
a.index("t")
a.index("k")  # 없으면 오류 발생
```

```python
## 문자열 삽입 - join()
",".join(["a", "b", "c", "d"])
```

```python
## 소문자를 대문자로 바꾸기 - upper()
a = "hi"
a.upper()
```

```python
a = "HI"
a.lower()
```

```python
## 양쪽 공백 지우기 - strip(), 중간 공백은 지워지지 않음
a = "  hi  "
a.strip()
```

```python
## 문자열 바꾸기 - replace()
a = "Life is too short"
a.replace("Life", "Leg")
a.replace("Life", "")
```

```python
## 문자열 나누기 - split()
a = "Life is too short"
print(a.split(" "))
print(a.split("i"))
## print(a.split("")) 오류남
print(a.split("k"))  # 없으면 전체 문자열 반환
```
