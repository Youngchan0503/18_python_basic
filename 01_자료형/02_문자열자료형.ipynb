```python
# 문자열은 어떻게 만들고 사용할까?

## 1. 큰 따옴표
print("hello world")

## 2. 작은 따옴표
print('python is fun')

## 3. 큰 따옴표 3개를 연속으로 감싸기
print("""Life is too short, You need python""")

## 4. 작은 따옴표 3개를 연속으로 감싸기
print('''Life is too short, You need python''')
```

```python
print("Chuls's favorite food is Kimchi")
print('Chuls's favorite food is Kimchi')
```

```python
# 문자열에 큰따옴표 표현하기
print('"python is ver fun" chulsu says')
```

```python
## 역슬래시 사용해서 작은따옴표, 큰따옴표 표시
print('Chuls\'s favorite food is Kimchi')
print("\"python is ver fun\" chulsu says")
```
