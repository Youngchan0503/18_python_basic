```python
# 딕셔너리는 어떻게 만들까
## {key1 : value1, key2 : value2, ...}

dict ={
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

type(dict)
```

```python
## 딕셔너리 쌍 추가하기
a ={1: "a"}
a[2] = "b"  # 2는 index가 아니라 key값

a
```

```python
a["name"] = "홍길동"
print(a)
```

```python
a[3] = [1, 2, 3]
a
```

```python
a = {
    1: "a",
    2: "b",
    "name": "홍길동",
    3: [1, 2, 3],
    (1,2): (1,2),
    # [1,2]: [1,2]   # 여기에서 문제, key가 리스트는 안됨, 수정할수 있는 거는 key로 지정 안됨
}
a
```

```python
## 딕셔너리에서 key를 사용해서 value 값 얻기
grade = {
    "홍길동": 10,
    "유재석": 90
}

print(grade["유재석"])

grade["홍길동"] = 80
print(grade)

grade["강호동"] = 70
print(grade)
```

```python
## 딕셔너리 관련 함수
## key리스트 만들기 - keys()

dict ={
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

print(dict.keys())
print(list(dict.keys()))
```

```python
## value 리스트 만들기 - values()
dict ={
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

print(dict.values())
print(list(dict.values()))
```

```python
## key, value 쌍 얻기 - items
dict ={
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

dict.items()
list(dict.items())
```

```python
## key와 value 쌍  모두 지우기 - clear()
dict ={
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

print(dict)
dict.clear()
print(dict)
```

```python
## key로 value값 얻기 - get()
dict ={
    "name": "kim",
    "phone": "010-1234-1234",
    "birth": "09-02"
}

print(dict.get("name"))
print(dict["name"]) # 둘다 같음

dict.get("hobby")
# dict["hobby"]  # 이거는 에러남
```
