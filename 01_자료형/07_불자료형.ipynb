```python
print(bool(1 == 1))
print(bool("python"))       #문자는 다 True
print(bool(""))             #빈 문자만 False
print(bool(10))
print(bool(0))              #숫자는 0만 False
print(bool([1, 2, 3]))
print(bool([]))             #비어있는 리스트는 False

## 그냥 비어있으면 False라고 생각하면 됨
```

```python
## 논리연산자

# and 
# or

print(True and True)
print(True and False)
print(False and True)
print(False and False)

print("="*5)

print(True or True)
print(True or False)
print(False or True)
print(False or False)



## 비교 연산자

# >         크다
# <         작다
# >=        크거나 같다, 이상
# <=        작거나 같다, 이하
# !=        다르다
# ==        같다
```
