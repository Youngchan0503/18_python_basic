```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, ...]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["Life", "is"]]
e = [1, 2, "life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```
