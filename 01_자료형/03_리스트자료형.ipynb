```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, ...]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["Life", "is"]]
e = [1, 2, "life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트의 연산
a = [1, 2, 3]
b = [4, 5, 6]
a + b
```

```python
a * 3
```

```python
## 리스트 길이 구하기 - len
a = [1,2,3]
len(a)
```

```python
## 리스트의 수정과 삭제
a = [1, 2, 3]
a[2] = 4
```

```python
a 
```

```python
a = [1, 2, 3]
del a[1]
print(a)
```

```python
## 리스트 관련 함수

## 리스트의 요소 추가하기 - append
a = [1, 2, 3]
a.append(4)
a
```

```python
a.append(["a", "b"])
a
```

```python
a = [1, 2, 3, 4, 5]

print(a[::2])
## print(a[4:0:-1]) #이거는 안됨
print(a[::-1]) #이거는 안됨
print(a[1:1]) #이거는 안됨 빈 리스트 나옴
```

```python
## 리스트 정렬 - sort
a = [1, 4, 3, 2]
a.sort()
a
```

```python
a = ["a", "c", "b"]
a.sort(reverse=True)
a
```

```python
## 리스트 뒤집기 - reverse
a = ["a", "c", "b"]
a.reverse()
a
```

```python
## 인덱스 반환 - index
a = [1, 2, 3, 4]
a.index(3)
```

```python
## 리스트에 요소 삽입 - insert 인덱스 위치에 삽입
a = [1, 2, 3]
a.insert(0, 4)
a.insert(len(a), 5)
a
```

```python
## 리스트 요소 제거 - remove()
a = [1, 2, 3, 1, 2, 3]
a.remove(2)  # 앞에 있는 2만 제거됨
a
```

```python
## 리스트 요소 꺼내기 - pop()
a = [1, 2, 3]
number = a.pop()  # 맨 마지막 요소를 돌려주고 그 요소는 삭제됨

print(number)
print(a)
```

```python
## 리스트에 포함된 요소 x의 개수 - count()
a = [1, 2, 3, 1]
a.count(1) 
```

```python
## 리스트 확장
a = [1, 2, 3]
a.extend([4, 5])
## append는 리스트 형태가 그대로 들어갔지만 extend는 리스트 안에 있는 요소들이 들어감
```
