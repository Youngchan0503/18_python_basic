```python
# 집합 연산 사용하기

## 합집합
## 세트1 | 세트2
## set.union(세트1, 세트2) => 세트1.union(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
## 교집합
## 세트1 & 세트2
## set.intersection(세트1, 세트2) => 세트1.intersection(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
## 차집합
## 세트1 - 세트2
## set.difference(세트1, 세트2) => 세트1.difference(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a, b))
print(a.difference(b))
print("=" * 20)
print(b -a)
```

```python
## 대칭 차집합
## 세트1 ^ 세트2
## set.symmetric_difference(세트1, 세트2) => 세트1.symmetric_difference(세트2)

print(a ^ b)
print(set.symmetric_difference(a, b))
print(a.symmetric_difference(b))
```

```python
# 집합 연산 후 할당 연산자 사용하기

# update
## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}
a |= {5}
print(a)

a = {1, 2, 3, 4}
a.update({5})
print(a)
```

```python
# intersection_update
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)

a = {1, 2, 3, 4}
a.symmetric_difference_update({3, 4, 5, 6})
print(a)
```

```python
# 부분집합과 상위집합 확인하기
## 현재세트 <= 다른세트
## 현재세트.issubset({다른세트})

a = {1, 2, 3, 4}
a <= {1, 2, 3, 4}
```

```python
a.issubset({1, 2, 3, 4})
```

```python
# 세트 조작하기

## 세트에 요소 추가하기
a = {1, 2, 3, 4}
a.add(5)
print(a)
```

```python
# 세트 표현식 사용하기
## {식 for 변수 in 반복가능한객체}
## set(식 for 변수 in 반복가능한객체)

a = {i for i in "apple"}
print(a)
```

```python
# 세트 표현식에 if 조건문 사용하기
a = {i for i in "pineapple" if i not in "apl"}
print(a)
```

```python
# 연습문제
a = {i for i in range(1,101) if i % 3 == 0}
b = {i for i in range(1,101) if i % 5 == 0}

print(a & b)
```

```python
# 심사문제
first, second = map(int, input().split())

a  = {i for i in range(1, first + 1) if first % i == 0}
b  = {i for i in range(1, second + 1) if second % i == 0}

divisor = a & b

result = 0
if type(divisor) == set:
    result = sum(divisor)

print(result)
```
