```python
# 딕셔너리 조작하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40,
}

print(x)

x.setdefault("e")
print(x)

x.setdefault("f", 100)
print(x)
```

```python
# 딕셔너리에서 키의 값 수정하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40,
}

x.update(a=90)
print(x)

x.update(e=50)
print(x)
```

```python
x.update(a=900, f=60)
print(x)
```

```python
y = {
    1: "one",
    2: "two"
}

y.update({
    1: "ONE", 
    3: "THREE"
})
print(y)
```

```python
y.update(
    [
        [2, "TWO"],
        [4, "FOUR"]
    ]
)

print(y)
```

```python
y.update(
    zip([1,2], ["one", "two"])
)

print(y)
```

```python
# 딕셔너리 표현식 사용하기
# {키: 값 for 키, 값 in 딕셔너리}

keys = ["a", "b", "c", "d"]

imsi = dict.fromkeys(keys)
print(imsi)

x = {key: value for key, value in imsi.items()}
print(x)
```

```python
{key: 0 for key in dict.fromkeys(["a", "b", "c", "b"]).keys()}
```

```python
{value: 0 for value in {"a": 10, "b": 20, "c": 30, "d": 40}.values()}
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40,
}

for key, value in x.items(): # 반복이 돌아가는 중에 크기가 바뀌면 다음 참조 위치가 잘못될수도 있으니 에러 <- del은 메모리에서 삭제하는거니까
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40,
}

y = x.copy()

for key, value in y.items():
    if value == 20:
        del x[key]

print(x)
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40,
}

x = {key: value for key, value in x.items() if value != 20}  # 필터링

print(x)
```

```python
# 심사문제
keys = input().split()
values = map(int, input().split())

x = dict(zip(keys, values))

x = {key: value for key, value in x.items()}
y = x.copy()

for key, value in y.items():
    if key == "delta" or value == 30:
        del x[key]


print(x)
```
