```python
# 딕셔너리 만들기
# 딕셔너리 = {
#     키1 : 값1,
#     키2 : 값2
# }

lux = {
    "health" : 490,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}

print(lux)
```

```python
# 키 이름이 중복이 되면?
lux = {
    "health" : 490,
    "health" : 800,
    "mana" : 334,
    "melee" : 550,
    "armor" : 18.72
}

print(lux["health"])
```

```python
# 딕셔너리의 키의 자료형
## 키에는 리스트와 딕셔너리는 사용할 수 없음, 둘다 안에 요소가 변경가능하기 때문에

x = {
    100 : "hundred",
    False : 0,
    3.5 : [3.5, 3.5]
}

print(x)
```

```python
x = {
    [10, 20] : 100
}
```

```python
# 빈 딕셔너리 만들기
x = {}
y = dict()

print(x)
print(y)

print(type(x))
print(type(y))
```

```python
# dict()로 딕셔너리 만들기
lux1 = dict(
    health=490,
    mana=334,
    melee=550,
    armor=18.72
)

print(lux1)
```

```python
iter1 = ["health", "mana", "melee", "armor"]
iter2 = [490, 334, 550, 18.72]
dict(zip(iter1,iter2))
```

```python
lux3 = dict(
    [('health',490), ('mana',334), ('melee', 550), ('armor', 18.72)]
)
print(lux3)
```

```python
lux4 = dict({'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72})
print(lux4)
```

```python
# 딕셔너리의 키에 접근하고 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print(lux["health"])
print(lux["armor"])
```

```python
# 딕셔너리의 키에 값 할당하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
lux["health"] = 2037
lux["mana"] = 1184
print(lux)
```

```python
lux["mana_regen"] = 3.28
print(lux)
```

```python
lux["attack_speed"]
```

```python
# 딕셔너리에 키가 있는지 확인하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
print("health" in lux)
print("attack_speed" in lux)
```

```python
# 딕셔너리의 키 개수 구하기
lux = {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
len(lux)
```
