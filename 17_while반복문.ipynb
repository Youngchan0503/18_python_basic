```python
# while 반복문 사용하기

# 초기식
# while 조건식:
#     반복할 코드
#     변화식

i = 0
while i < 10:
    print("Hello world!")
    i += 1
```

```python
# 초깃값을 1부터 시작하기
i = 1
while i <= 5:
    print("hellow world", i)
    i += 1
```

```python
# 초깃값을 감소시키기
i = 5
while i > 0:
    print("hello", i)
    i -= 1
```

```python
count = int(input())

i = 0
while i < count:
    print("hello", i)
    i += 1
```

```python
# 연습문제
i = 2
j = 5

while i <= 32 or j > 0:
    print(i, j)
    i *= 2
    j -= 1
```

```python
# 심사문제
money = int(input())

while money >= 1350:
    money -= 1350
    print(money)
```
