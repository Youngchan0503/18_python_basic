```python
# 중첩 루프 사용하기
for i in range(5):
    for j in range(5):
        print("j:", j, sep="", end=" ")
    print("i:", i, "\\n", sep="")
```

```python
# 연습문제

for i in range(5):
    for j in range(5):
        if i <= j:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 심사문제
height = int(input())
for i in range(height):
    print(" " * (height - i -1 ), end="")
    print("*" * ((i * 2) +1), end="")
    print()
```

```python
# 심사문제 (2중 for문으로 한다면)
height = int(input())

for i in range(height):
    for j in range(height):
        if j + i < height -1:
            print(" ", end="")
        else:
            print("*", end="")
    print("*" * i)
```
