```python
# 1부터 100까지 숫자 출력하기
for i in range(1, 11):
    print(i)
```

```python
# 3의 배수일때 Fizz, 5의 배수일때 Buzz
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 코드 단축하기
for i in range(1, 31):
    print("Fizz" * (i % 3 == 0) + "Buzz" * (i % 5 == 0) or i)
```

```python
# 연습문제
for i in range(1, 101):
    if i % 2 == 0 and i % 11 == 0:
        print("FizzBuzz")
    elif i % 2 == 0:
        print("Fizz")
    elif i % 11 == 0:
        print("Buzz")
    else:
        print(i)
```

```python
# 심사문제
first, second = map(int, input().split())

for i in range(first, second + 1):
    print("Fizz" * (i % 5 == 0) + "Buzz" * (i % 7 == 0) or i)
```

```python
# 문제: 중첩 for를 이용
test = [
    [1, 3, 5, 7, 9],
    ["a" "b", "c", "d", "e"],
    ["1", "2", "3", "4", "5"]
]

for i in test:
    for j in range(len(i)):
        print(i[j], end="")
    print(end=" ")

print("\n-------------------------")
for i in test:
    for j in i:
        print(j, end="")
    print(end=" ")
```
